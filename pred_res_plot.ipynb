```python
import pandas as pd
import os
import matplotlib.pyplot as plt
from collections import Counter
import numpy as np
```

```python
src_ = "./../src/"
test_mono_path = os.path.join(src_, "HKSL_Prediction_Cynthia_test_mono.xlsx")
test_poly_path = os.path.join(src_, "HKSL_Prediction_Cynthia_test_poly.xlsx")
```

```python
# df = pd.read_excel(test_mono_path)
df = pd.read_excel(test_poly_path)
```

```python
def func(pct, allvals):
    absolute = int(np.round(pct/100.*np.sum(allvals)))
    return f"{pct:.1f}%\n({absolute:d})"
```

```python
def plot_ratio_pie(df, name, range_start, range_end, side="Right"):
    df = df[df["sign_name"] == name]
    df = df[df["side"] == side]
    df = df[(df["frame_idx"] >= range_start - 1) & (df["frame_idx"] < range_end - 1)]
    preds = df["predicted"].tolist()
    pred_counts = Counter(preds)
    sizes = list(pred_counts.values())
    labels = list(pred_counts.keys())
    print(sizes)
    # Create a pie chart
    plt.figure(figsize=(8, 8))  # Adjust the size of the figure
    plt.pie(sizes, labels=labels, autopct=lambda pct: func(pct, sizes), startangle=140)
    plt.title(f'Predictions [{df["target"].tolist()[0]}]')
    plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

    # Display the pie chart
    plt.show()
```

```python
plot_ratio_pie(df, "HKSL_lesson_only54-DRAGON_BOAT_FESTIVAL-0UNF-516", 20, 42)
```
