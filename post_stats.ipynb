```python
import pandas as pd
from paths import *
```

```python
filepath = src_dir + "HKSL_Prediction_Cynthia_train.xlsx"
# Read the Excel file
df = pd.read_excel(filepath)
filtered_df = df[(df['side'] == 'Right') & (df['missing_detection'] == 1)]
```

```python
new_filtered_df = filtered_df.drop_duplicates(subset='sign_name', keep='first', inplace=False)
```

```python
filtered_df
```

```python
new_filtered_df
```

```python
target_counts = new_filtered_df['target'].value_counts()
print(target_counts)
```

```python
target_counts = filtered_df['target'].value_counts()
print(target_counts)
```

```python
target_counts_df = pd.DataFrame(target_counts)
target_counts_df = target_counts_df.rename(columns={'target': 'count'})
target_counts_df.index.name = 'target'
target_counts_df.reset_index(inplace=True)
target_counts_df
```

```python
target_counts_df['unmarked'] = target_counts_df['target'].apply(lambda x: 1 if x in ['B', 'y', '2', 'C', '6', '>'] else 0)
```

```python
target_counts_df
```

```python
correlation = target_counts_df['count'].corr(target_counts_df['unmarked'])
correlation
```
